```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv(r"C:\Users\Bhoomika\rental_data_cleaned.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df[["area", "beds", "bathrooms", "balconies", "rent"]].describe()
```

```python
city_analysis = (
    df.groupby("city")
      .agg(
          Total_Properties=("rent", "count"),
          Average_Rent=("rent", "mean"),
          Median_Rent=("rent", "median")
      )
      .round(2)
      .sort_values("Average_Rent", ascending=False)
)

print(city_analysis)
```

```python
city_analysis["Average_Rent"].sort_values().plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Average Monthly Rent by City")
plt.xlabel("City")
plt.ylabel("Average Rent (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
property_analysis = (
    df.groupby("property_type")
      .agg(
          Property_Count=("rent", "count"),
          Average_Rent=("rent", "mean"),
          Median_Rent=("rent", "median")
      )
      .round(2)
      .sort_values("Average_Rent", ascending=False)
)

print(property_analysis)
```

```python
df["property_type"].value_counts().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Number of Listings by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Number of Listings")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
property_analysis["Average_Rent"].sort_values().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Monthly Rent by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Average Rent (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["rent"].dropna(), bins=50)

plt.title("Distribution of Monthly Rental Prices")
plt.xlabel("Monthly Rent (₹)")
plt.ylabel("Number of Properties")

plt.tight_layout()
plt.show()
```

```python
print("Average rent:", round(df["rent"].mean(), 2))
print("Median rent:", round(df["rent"].median(), 2))
print("Maximum rent:", df["rent"].max())
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["rent"].dropna(), bins=50)
plt.yscale("log")

plt.title("Rental Price Distribution (Logarithmic Frequency Scale)")
plt.xlabel("Monthly Rent (₹)")
plt.ylabel("Number of Properties (Log Scale)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["area"],
    df["rent"],
    alpha=0.5
)

plt.title("Property Area vs Monthly Rent")
plt.xlabel("Area (Square Feet)")
plt.ylabel("Monthly Rent (₹)")

plt.tight_layout()
plt.show()
```

```python
correlation = df["area"].corr(df["rent"])

print("Correlation between area and rent:", round(correlation, 3))
```

```python
furnishing_analysis = (
    df.groupby("furnishing")
      .agg(
          Property_Count=("rent", "count"),
          Average_Rent=("rent", "mean"),
          Median_Rent=("rent", "median")
      )
      .round(2)
      .sort_values("Average_Rent", ascending=False)
)

print(furnishing_analysis)
```

```python
furnishing_analysis["Average_Rent"].sort_values().plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Average Rent by Furnishing Category")
plt.xlabel("Furnishing")
plt.ylabel("Average Monthly Rent (₹)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
bedroom_analysis = (
    df.groupby("beds")
      .agg(
          Property_Count=("rent", "count"),
          Average_Rent=("rent", "mean"),
          Median_Rent=("rent", "median")
      )
      .round(2)
      .sort_index()
)

print(bedroom_analysis)
```

```python
bedroom_analysis["Average_Rent"].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Average Rent by Number of Bedrooms")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Monthly Rent (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df["rent_per_sqft"] = (
    df["rent"] / df["area"].replace(0, np.nan)
)
```

```python
rent_per_sqft_analysis = (
    df.groupby("city")["rent_per_sqft"]
      .mean()
      .round(2)
      .sort_values(ascending=False)
)

print(rent_per_sqft_analysis)
```

```python
rent_per_sqft_analysis.sort_values().plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Average Rent per Square Foot by City")
plt.xlabel("City")
plt.ylabel("Average Rent per Sq. Ft. (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
